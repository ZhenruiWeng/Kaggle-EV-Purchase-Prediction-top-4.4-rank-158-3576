```python
import time

T0 = time.time()
STAGES = 12


def pit(stage, label):
    """A compact car with an exhaust trail drives stage/STAGES of the way to the flag."""
    width = 60
    pos = int((width - 7) * stage / STAGES)
    trail = "° ° °"[-min(pos, 5):] if pos else ""
    print(f"\n LAP {stage:2d}/{STAGES} · {label} · {time.time() - T0:6.1f}s")
    print(" " + " " * pos + " _/=\\_ " + " " * (width - pos - 7) + " |#_#|")
    print(" " + "=" * (pos - len(trail)) + trail + "'o---o'" + "-" * (width - pos - 7) + " |_#_|")


def box(title, lines, width=82):
    """Draw text inside an ASCII box."""
    print("+" + "-" * (width - 2) + "+")
    print("| " + title.upper().ljust(width - 4) + " |")
    print("+" + "-" * (width - 2) + "+")
    for line in lines:
        print("| " + line.ljust(width - 4) + " |")
    print("+" + "-" * (width - 2) + "+")


def strip(values, lo, hi, width=48):
    """Dot strip on a fixed axis: more dots in one cell -> a taller glyph."""
    cells = [0] * width
    for v in values:
        cells[min(max(int((v - lo) / (hi - lo) * (width - 1)), 0), width - 1)] += 1
    return "".join(" .:|#"[min(c, 4)] for c in cells)


def podium(first, second, third):
    """Three finishers, each a (label, detail) tuple."""
    print("\n".join([
        r"                          \o/",
        r"                        .-----.",
        r"            \o/         |  1  |",
        r"          .-----.       |     |         \o/",
        r"          |  2  |       |     |       .-----.",
        r"          |     |       |     |       |  3  |",
        r"   =========================================================",
    ]))
    for place, (label, detail) in zip((1, 2, 3), (first, second, third)):
        print(f"      P{place}  {label:31s} {detail}")


TROPHY = r"""
                 ___________
                '._==_==_=_.'        CHECKERED FLAG!  THE LAST ONE.
                .-\:      /-.        |#_#_#_#_#_#_#_#|
               | (|: V6   |) |       |_#_#_#_#_#_#_#_|
                '-|:      |-'        |#_#_#_#_#_#_#_#|
                  \::.    /          |
                   '::. .'           |        _/=\_
                     ) (             |   ° ° 'o---o'  >>>  thanks for racing
                   _.' '._           |
                  '-------'     =====|=======================
"""


pit(0, "grid position, engines warming up")
```

🎓 **Pit-stop lesson: check the alignment of every test file with a correlation fingerprint.** OOF files come with
labels you can check. Test files don't, and one shifted or reordered test file silently ruins a submission. Two
engines that agree on the train rows must agree about as much on the test rows. So for every engine we compare
its correlation with a reference engine on OOF and on test. The check is **one-sided**: test predictions are
averages of the fold models, so they are smoother and correlate slightly *more* (up to +0.016 here for noisy
neural nets). A misaligned file correlates *less*. Shuffling just 1% of its rows costs −0.01, and a one-row
shift drops the correlation to about zero. An engine whose test correlation falls more than 0.005 below its OOF
correlation is dropped.

## Lap 3 · One track for every engine: rank → probit, and the fingerprint check

ROC AUC only asks **"is this positive ranked above that negative?"**. Any strictly increasing transform of a
prediction leaves its AUC unchanged. The engines, however, arrive on very different scales: probabilities,
percentile ranks and even raw margins such as `-3.17`.

We therefore replace every column by its rank and push the rank through the inverse normal CDF:

$$g = \Phi^{-1}\!\left(\frac{\operatorname{rank}(s)}{n+1}\right)$$

🎓 **Pit-stop lesson: blend in probit space.** Plain rank averaging treats the middle of the ranking like its
tails. Probit ranks behave like logits: confident rows get room to disagree, which usually helps AUC a little.
Train and test are transformed separately, because each is ranked within its own set.

```python
def probit(values):
    return norm.ppf(rankdata(values) / (len(values) + 1)).astype(np.float32)


NAMES = list(OOF)
P_OOF = np.column_stack([probit(OOF[n]) for n in NAMES])
P_TEST = np.column_stack([probit(TEST[n]) for n in NAMES])
del OOF, TEST

# correlation fingerprint: corr(engine, reference) must match on OOF and on test
REF_NAME = "fx_fusion"
r_ix = NAMES.index(REF_NAME)
fingerprint = pd.DataFrame({
    "oof_corr": [np.corrcoef(P_OOF[:, i], P_OOF[:, r_ix])[0, 1] for i in range(len(NAMES))],
    "test_corr": [np.corrcoef(P_TEST[:, i], P_TEST[:, r_ix])[0, 1] for i in range(len(NAMES))]}, index=NAMES)
fingerprint["gap"] = fingerprint.test_corr - fingerprint.oof_corr
bad = fingerprint.index[fingerprint.gap < -0.005].tolist()
print(f"fingerprint (test − OOF correlation with {REF_NAME}): min {fingerprint.gap.min():+.5f} ({fingerprint.gap.idxmin()})"
      f" · median {fingerprint.gap.median():+.5f} · max {fingerprint.gap.max():+.5f} ({fingerprint.gap.idxmax()})"
      f" · dropped: {bad or 'none'}")
keep = [i for i, n in enumerate(NAMES) if n not in bad]
NAMES = [NAMES[i] for i in keep]
P_OOF, P_TEST = P_OOF[:, keep], P_TEST[:, keep]


def fast_auc(labels, score):
    """Mann-Whitney AUC in one argsort (ties are negligible for continuous blends)."""
    order = np.argsort(score, kind="quicksort")
    ranked = labels[order]
    n_pos = ranked.sum()
    n_neg = len(ranked) - n_pos
    return (np.flatnonzero(ranked).sum() + n_pos - n_pos * (n_pos + 1) / 2) / (n_pos * n_neg)


assert abs(fast_auc(yb, P_OOF[:, 0]) - roc_auc_score(y, P_OOF[:, 0])) < 1e-6
pit(3, f"{len(NAMES)} engines on the probit track")
```

## Lap 4 · The AUC Grand Prix

Each engine races alone. The distance covered is its OOF AUC. The track starts at 0.9455 and ends at the pole car;
engines below the start line wait in the pit lane. Weak cars can still matter later: a car that takes a different
line through the corners can help a team.

```python
single_auc = pd.Series({n: fast_auc(yb, P_OOF[:, i]) for i, n in enumerate(NAMES)}).sort_values(ascending=False)
START = 0.9455
lo, hi = START, single_auc.max()
print(f"{'engine':22s} {'track (start 0.9455 -> pole)':48s}  OOF AUC")
for name, score in single_auc.items():
    if score < lo:
        continue
    dist = int(44 * (score - lo) / (hi - lo))
    print(f"{name[:22]:22s} {'.' * dist}o=o>{' ' * (44 - dist)}  {score:.6f}")
pit_lane = single_auc[single_auc < lo]
print(f"\n[pit lane] {len(pit_lane)} engines below the start line, best {pit_lane.index[0]} {pit_lane.iloc[0]:.6f},"
      f" slowest {pit_lane.index[-1]} {pit_lane.iloc[-1]:.6f}")
BEST_SINGLE = single_auc.index[0]
print(f"🏆 pole position: {BEST_SINGLE} · {single_auc.iloc[0]:.6f}")
pit(4, "Grand Prix finished")
```

## Lap 5 · Same slipstream, same car?

Two engines with rank correlation 0.999 are nearly the same car in different paint. Counting them twice gives
their shared errors a double vote. We cluster the engines by correlation (average linkage, cut at distance
0.002, i.e. ρ ≈ 0.998) and pick each cluster's **medoid**, the member most correlated with its teammates.
Picking the medoid uses no labels.

🎓 **Pit-stop lesson: count independent orderings, not files.** Correlation between the top public files
is 0.996–0.9999. Diversity is the only fuel left, and it is scarce.

```python
corr = np.corrcoef(P_OOF.T)
dist = np.clip(1 - corr, 0, 2)
np.fill_diagonal(dist, 0)
labels = fcluster(linkage(squareform(dist, checks=False), method="average"), t=0.002, criterion="distance")
clusters = {}
for name, lab in zip(NAMES, labels):
    clusters.setdefault(int(lab), []).append(name)
medoids = []
for lab, members in sorted(clusters.items()):
    ix = [NAMES.index(m) for m in members]
    medoid = members[int(np.argmax(corr[np.ix_(ix, ix)].mean(1)))]
    medoids.append(medoid)
    if len(members) > 1:
        print(f"cluster {lab:2d} · {len(members):2d} cars · medoid {medoid:22s} · {', '.join(members)}")
print(f"... plus {sum(len(m) == 1 for m in clusters.values())} solo drivers")
print(f"\n{len(NAMES)} engines → {len(medoids)} independent voices")
pit(5, "slipstreams mapped")
```

## Lap 6 · Nested Caruana hill climb

[Caruana et al. (2004)](https://www.cs.cornell.edu/~caruana/ctp/ct.papers/caruana.icml04.icdm06long.pdf):
start from an empty blend and repeatedly add the engine (with replacement) that raises AUC the most. Stop when
nothing helps. The integer counts become the weights.

The trap: if you pick the engines on all 668k rows and then report the AUC on the *same* rows, the score is
optimistic. So the climb runs **inside** each of 5 folds. It selects on 4/5 of the rows and is scored on the
held-out fifth. Only if the nested blend beats the pole-position car on **at least 4 of 5 folds** do we refit
the climb on all rows and use it.

To save CPU the climb only considers engines past OOF AUC 0.9450. Offline, adding the slower cars changed
nothing: the climb never picked one.

🎓 **Pit-stop lesson: the thing that chooses must be validated too.** Model selection is a model.

```python
def hill_climb(P, labels, candidates, max_steps=80, tol=1e-7):
    blend = np.zeros(len(P), dtype=np.float64)
    counts = np.zeros(P.shape[1])
    best = -1.0
    for _ in range(max_steps):
        scores = {j: fast_auc(labels, blend + P[:, j]) for j in candidates}
        j = max(scores, key=scores.get)
        if scores[j] <= best + tol:
            break
        best = scores[j]
        blend += P[:, j]
        counts[j] += 1
    return counts / counts.sum(), best


CANDIDATES = [i for i, n in enumerate(NAMES) if single_auc[n] >= 0.9450]
folds = list(StratifiedKFold(N_FOLDS, shuffle=True, random_state=SEED).split(P_OOF, y))
ref = NAMES.index(BEST_SINGLE)
hill_oof = np.zeros(len(y))
fold_rows = []
for k, (tr, va) in enumerate(folds):
    w, _ = hill_climb(P_OOF[tr], yb[tr], CANDIDATES)
    hill_oof[va] = P_OOF[va] @ w
    a_hill, a_ref = fast_auc(yb[va], hill_oof[va]), fast_auc(yb[va], P_OOF[va, ref])
    fold_rows.append({"fold": k, "best_single": a_ref, "nested_hill": a_hill, "delta": a_hill - a_ref,
                      "engines_used": int((w > 0).sum())})
    print(f"  fold {k}: hill {a_hill:.6f} vs {BEST_SINGLE} {a_ref:.6f} · Δ {a_hill - a_ref:+.6f}"
          f" · {int((w > 0).sum())} engines")
fold_table = pd.DataFrame(fold_rows)
wins = int((fold_table.delta > 0).sum())
HILL_ACCEPTED = wins >= 4 and fold_table.delta.mean() > 0
print(f"nested hill OOF {fast_auc(yb, hill_oof):.6f} · wins {wins}/5 · accepted: {HILL_ACCEPTED}"
      f" · {len(CANDIDATES)} candidates")

W_FULL, full_auc = hill_climb(P_OOF, yb, CANDIDATES)
weights = pd.Series(W_FULL, index=NAMES)
weights = weights[weights > 0].sort_values(ascending=False)
print("\nfull-data hill climb (in-sample AUC, optimistic): %.6f" % full_auc)
for name, wt in weights.items():
    print(f"  {name:22s} {'#' * int(round(wt * 40)):40s} {wt:.3f}")
pit(6, "hill climbed")
```

## Lap 7 · Control cars and the team car

A result means little without controls. Each car below uses the same 5 folds:

* **equal mean** of every engine past the 0.9455 start line;
* **medoid mean**, one vote per correlation cluster (Lap 5), again only for cars past the start line;
* **logistic stack**, an L2 logistic regression on all probit columns (a learned linear blend).

Then the **team car**: three quarters hill climb, one quarter logistic stack (each standardized first). The two
combiners fail in different ways. The hill climb uses a handful of engines with coarse integer weights, while the
stack spreads small weights over all of them. The team car must beat the hill climb alone on at least 4 of 5
folds, or it stays parked.

Offline I also tried these cars. They stay in the pit lane to save CPU time:

| car | nested OOF AUC | verdict |
|---|---:|---|
| LightGBM stacker on probits + raw features | 0.946362 | loses to the best single engine |
| label-based Fisher / LDA, ridge 0–0.2 | 0.945168–0.946419 | unstable with dozens of near-collinear inputs |
| hill climb that may also *subtract* engines | +0.000008 over plain hill | only 4/5 folds |
| "mined" pure-label income/commute intervals as extra rules | −0.00006 to −0.00039 | noise: only the big known rules survive (Lap 9) |
| sometimessubodh's lasso meta-stack as an engine | never picked | it is a stack of engines already here |
| finer hill climb (no tolerance, 200 steps) or top-3 warm start | ±0.000001 | same blend |

🎓 **Pit-stop lesson: with 0.99+ correlated inputs, flexible stackers mostly learn noise.** Few parameters and
a hard validation gate usually win. A *small* share of a regularized stack can still help, because its errors
differ from the hill climb's.

```python
controls = {}
strong = [i for i, n in enumerate(NAMES) if single_auc[n] >= START]
controls["equal mean (past start line)"] = P_OOF[:, strong].mean(1)
controls["medoid mean (past start line)"] = P_OOF[:, [NAMES.index(m) for m in medoids if single_auc[m] >= START]].mean(1)
stack_oof = np.zeros(len(y))
for tr, va in folds:
    lr = LogisticRegression(C=0.01, max_iter=400).fit(P_OOF[tr], y[tr])
    stack_oof[va] = lr.decision_function(P_OOF[va])
controls["logistic stack"] = stack_oof
controls["nested hill climb"] = hill_oof


def zscore(v):
    return (v - v.mean()) / v.std()


team_oof = 0.75 * zscore(hill_oof) + 0.25 * zscore(stack_oof)
controls["team car: hill + 1/4 stack"] = team_oof
controls[f"pole: {BEST_SINGLE}"] = P_OOF[:, ref]
race = pd.DataFrame([{"car": k, "oof_auc": fast_auc(yb, v),
                      "fold_wins_vs_pole": sum(fast_auc(yb[va], v[va]) > fast_auc(yb[va], P_OOF[va, ref]) for _, va in folds)}
                     for k, v in controls.items()]).sort_values("oof_auc", ascending=False)
lo, hi = race.oof_auc.min(), race.oof_auc.max()
for row in race.itertuples():
    dist = int(40 * (row.oof_auc - lo) / max(hi - lo, 1e-12))
    print(f"{row.car[:29]:29s} {'.' * dist}o=o>{' ' * (40 - dist)} {row.oof_auc:.6f}  wins {row.fold_wins_vs_pole}/5")
team_gain = [fast_auc(yb[va], team_oof[va]) - fast_auc(yb[va], hill_oof[va]) for _, va in folds]
TEAM_ACCEPTED = sum(g > 0 for g in team_gain) >= 4 and np.mean(team_gain) > 0
print()
print("team car vs hill climb per fold:", " ".join(f"{g:+.7f}" for g in team_gain), f"· accepted: {TEAM_ACCEPTED}")
own_oof = team_oof if TEAM_ACCEPTED else hill_oof
stack_full = LogisticRegression(C=0.01, max_iter=400).fit(P_OOF, y)   # the stack used on the test rows
pit(7, "control cars parked, team car chosen")
```

## Lap 8 · Does the gain beat the noise? Paired DeLong interval

Our blend and the pole car score **the same rows**, so their AUC errors are strongly correlated. DeLong's
method estimates the variance of the *difference* from each positive's and negative's contribution to the AUC.
This is much tighter than comparing two separate confidence intervals.

🎓 **Pit-stop lesson: a +0.00005 gain is only interesting with an interval.** Public LB noise on about 57k rows
is roughly ±0.001, which is 20× larger than the gains we are chasing.

```python
def delong_paired(labels, a, b):
    pos, neg = labels, ~labels
    m, n = pos.sum(), neg.sum()
    comps = []
    for s in (a, b):
        pooled = rankdata(s)
        v10 = (pooled[pos] - rankdata(s[pos])) / n
        v01 = 1 - (pooled[neg] - rankdata(s[neg])) / m
        comps.append((v10, v01))
    s10 = np.cov(np.vstack([comps[0][0], comps[1][0]]))
    s01 = np.cov(np.vstack([comps[0][1], comps[1][1]]))
    var = (s10[0, 0] + s10[1, 1] - 2 * s10[0, 1]) / m + (s01[0, 0] + s01[1, 1] - 2 * s01[0, 1]) / n
    return comps[0][0].mean() - comps[1][0].mean(), np.sqrt(max(var, 0))


delta, se = delong_paired(yb, own_oof, P_OOF[:, ref])
ci = (delta - 1.96 * se, delta + 1.96 * se)
print(f"our blend − {BEST_SINGLE}: Δ {delta:+.6f} · 95% CI [{ci[0]:+.6f}, {ci[1]:+.6f}] · z = {delta / se:.2f}")
pit(8, "interval measured")
```

## Lap 9 · Rules of the road: the generator's hard blocks

The training data was produced by a generator with a few **hard rules**. In 668k training rows these cells
are perfectly pure: every row in the first one bought an EV, and nobody in the other three did. The community
found them in the data (see megayak's and defiaudit's public-split probes, where inverting a single rule cost up
to −0.0126 public AUC):

| rule | direction |
|---|---|
| income ≥ 170,537 | always buys → top of the ranking |
| income 31,004 – 41,970 | never buys → bottom |
| daily commute ≥ 83 km | never buys → bottom |
| income exactly 30,000, no subsidy, and concern level 1 or Medium/High anxiety | never buys → bottom |

Tree ensembles *almost* learn these blocks, but smoothing and blending leak a few rows out of them. Below you'll
see rows inside a "never buys" cell ranked above 70% of the data. We push each cell to the very top or bottom
and keep the order inside the cell.

🎓 **Pit-stop lesson: post-process with rules only when the evidence is overwhelming.** I also tried *mining*
pure-label income intervals automatically. Validated fold by fold, that lost 0.00006–0.00039 AUC, because small
"pure" intervals are just luck. Four big, published, probe-confirmed rules pass the same test.

```python
def rule_masks(frame):
    income = frame["Annual_Income_USD"].to_numpy(np.float64)
    commute = frame["Daily_Commute_km"].to_numpy(np.float64)
    no_subsidy = frame["Subsidy_Available"].astype(str).to_numpy() == "No"
    concern_1 = frame["Environmental_Concern_Level"].to_numpy() == 1
    anxious = np.isin(frame["Range_Anxiety_Level"].astype(str).to_numpy(), ["Medium", "High"])
    return [
        ("income >= 170,537 -> always buys", income >= 170537, +1),
        ("income 31,004-41,970 -> never buys", (income >= 31004) & (income <= 41970), -1),
        ("commute >= 83 km -> never buys", commute >= 83, -1),
        ("30k, no subsidy, concern 1 or anxious", (income == 30000) & no_subsidy & (concern_1 | anxious), -1),
    ]


def apply_rules(score, frame):
    """Move each rule cell to the top (+1) or bottom (-1) block; order inside a cell is kept."""
    r = rankdata(score) / len(score)
    for _, mask, sign in rule_masks(frame):
        r[mask] += 2.0 * sign
    return rankdata(r) / len(r)


own_rank = rankdata(own_oof) / len(own_oof)
print(f"{'rule':40s} {'train rows':>10s} {'buyers':>7s} {'rate':>6s}   blend rank inside the cell (min/median/max)")
for label, mask, sign in rule_masks(train):
    print(f"{label:40s} {mask.sum():10,d} {y[mask].sum():7,d} {y[mask].mean():6.3f}   "
          f"{own_rank[mask].min():.4f} / {np.median(own_rank[mask]):.4f} / {own_rank[mask].max():.4f}")

rule_rows = []
for k, (_, va) in enumerate(folds):
    before = fast_auc(yb[va], own_oof[va])
    after = fast_auc(yb[va], apply_rules(own_oof[va], train.iloc[va]))
    rule_rows.append({"fold": k, "before": before, "after": after, "delta": after - before})
rule_table = pd.DataFrame(rule_rows)
RULES_ACCEPTED = bool((rule_table.delta >= 0).all() and rule_table.delta.mean() > 0)
print("\nper-fold gain from the rules:", " ".join(f"{d:+.7f}" for d in rule_table.delta),
      f"· mean {rule_table.delta.mean():+.7f} · accepted: {RULES_ACCEPTED}")
ruled_oof = apply_rules(own_oof, train) if RULES_ACCEPTED else own_rank
OWN_OOF = fast_auc(yb, ruled_oof)
print(f"our blend OOF {fast_auc(yb, own_oof):.6f} → with rules {OWN_OOF:.6f}")
pit(9, "rules of the road applied")
```

## Lap 10 · A pit-stop model calibrated on labels, not on the leaderboard

The public leader is jazivxt's `submission_latest_best.csv` (**public 0.94657**), a rank blend of hundreds of
scored public outputs. It has no OOF, so labels can't weight it directly. How much of it should we keep?

**The Fisher model** (idea: [thisray](https://www.kaggle.com/code/thisray/s6e9-fisher-blend-from-leaderboard-scores)).
Under a binormal model with Gaussianized scores, a file with AUC $a$ has class separation
$\delta = z/\sqrt{1+\pi(1-\pi)z^2}$ with $z = \sqrt2\,\Phi^{-1}(a)$. With within-class covariance
$\Sigma_w = \Sigma - \pi(1-\pi)\,\delta\delta^\top$, the blend $w\,g_1 + (1-w)\,g_2$ has

$$\text{AUC}(w) = \Phi\!\left(\frac{v^\top\delta}{\sqrt{2\,v^\top\Sigma_w v}}\right),\qquad v = (w,\;1-w).$$

The only unknown is the correlation $\rho$ in $\Sigma$. Version 1 plugged in the plain correlation of the two
files on the test rows and predicted a public score of 0.9468. The leaderboard said 0.94655. Version 2 fitted
$\rho$ to public scores. Version 3 asks **why** the plain correlation fails, and answers on labeled data.

**The experiment.** Take pairs of strong engines. For each pair, measure the real AUC of the 50/50 blend on OOF.
Then solve for the correlation $\rho_\text{eff}$ that makes the Fisher model reproduce it. The plain correlation is
far too low, because it is dominated by thousands of easy negatives that every model ranks the same way. The
correlation **inside the top half of the ranking**, where positives and hard negatives meet, tracks
$\rho_\text{eff}$ with an almost constant factor:

$$1-\rho_\text{eff} \;\approx\; k\,\bigl(1-\rho_\text{top 50\%}\bigr).$$

The cell below estimates $k$ from 66 labeled pairs and shows how stable it is.

🎓 **Pit-stop lesson: when a closed form fails, find the missing constant on data where you know the truth.**
Then check it against the few measurements you have, instead of fitting to them.

```python
def separation(auc, pi=PRIOR_POS_RATE):
    z = np.sqrt(2) * norm.ppf(auc)
    return z / np.sqrt(1 + pi * (1 - pi) * z * z)


def blend_auc(w, rho, a1, a2, pi=PRIOR_POS_RATE):
    d = np.array([separation(a1), separation(a2)])
    S_w = np.array([[1.0, rho], [rho, 1.0]]) - pi * (1 - pi) * np.outer(d, d)
    v = np.array([w, 1.0 - w])
    return norm.cdf(v @ d / np.sqrt(2.0 * v @ S_w @ v))


def top_half_corr(g1, g2):
    top = rankdata(g1 + g2) > len(g1) / 2
    return np.corrcoef(g1[top], g2[top])[0, 1]


# 12 strong engines that are not near-copies of each other -> 66 labeled pairs
pair_pool = []
for name in single_auc.index:
    i = NAMES.index(name)
    if all(corr[i, NAMES.index(p)] < 0.9995 for p in pair_pool):
        pair_pool.append(name)
    if len(pair_pool) == 12:
        break
cal_rows = []
for n1, n2 in combinations(pair_pool, 2):
    g1, g2 = P_OOF[:, NAMES.index(n1)].astype(np.float64), P_OOF[:, NAMES.index(n2)].astype(np.float64)
    a1, a2, a_mix = single_auc[n1], single_auc[n2], fast_auc(yb, g1 + g2)
    try:
        rho_eff = brentq(lambda r: blend_auc(0.5, r, a1, a2) - a_mix, 0.9, 0.99999999)
    except ValueError:
        continue
    rho_plain, rho_top = np.corrcoef(g1, g2)[0, 1], top_half_corr(g1, g2)
    cal_rows.append({"pair": f"{n1} / {n2}", "rho_plain": rho_plain, "rho_top50": rho_top, "rho_eff": rho_eff,
                     "k_plain": (1 - rho_eff) / (1 - rho_plain), "k_top50": (1 - rho_eff) / (1 - rho_top)})
cal = pd.DataFrame(cal_rows)
K_TOP = float(cal.k_top50.median())
for col in ("k_plain", "k_top50"):
    print(f"{col:8s}: median {cal[col].median():.3f} · IQR {cal[col].quantile(.25):.3f}–{cal[col].quantile(.75):.3f}"
          f" · relative spread {cal[col].std() / cal[col].mean():.1%}")
hi_k = float(np.ceil(cal.k_plain.quantile(0.98) * 10) / 10)
print(f"\nk = (1 − ρ_eff) / (1 − ρ_proxy) for each of the {len(cal)} labeled pairs, on one axis from 0 to {hi_k:.1f}:")
print(f"  from the plain correlation   |{strip(cal.k_plain, 0, hi_k)}|  spread {cal.k_plain.std() / cal.k_plain.mean():.0%}")
print(f"  from the top-half correlation|{strip(cal.k_top50, 0, hi_k)}|  spread {cal.k_top50.std() / cal.k_top50.mean():.0%}")
print(f"                               0{' ' * 22}{hi_k / 2:.2f}{' ' * 19}{hi_k:.1f}")
print("  A useful proxy piles up in one spot. The top-half correlation does; the plain one smears.")
print(f"\n{len(cal)} labeled pairs → k = {K_TOP:.3f}. Five examples:")
print(cal.sort_values("rho_eff").iloc[::max(len(cal) // 5, 1)][["pair", "rho_plain", "rho_top50", "rho_eff"]]
      .to_string(index=False, float_format=lambda v: f"{v:.6f}"))
pit(10, "pit-stop model calibrated on labels")
```

## Lap 11 · The timing tower: every file with a public score

**First, validation against the leaderboard.** Versions 1–3 of this notebook submitted three challengers, and every
engine they used is still in the garage, so the cell rebuilds all three. The calibrated pit-stop model then predicts
their public scores. It gets the anchor's public AUC and our own blend's public AUC, but it never sees the challenger
scores it is asked to predict. (The uncalibrated v1 model predicted **0.94680** for the first one.)

**Our own blend's public AUC is an estimate:** v1's measured 0.94648 plus the part of the OOF gain since v1 that the
board pays. Along this notebook's lineage (Fusion II → Fusion III → v1) the board paid 59% of OOF gains, and the
single engines below give a similar slope. (Mamarin's ledger found that it often pays nothing for blending gains.)

**The timing tower.** Versions 1–3 merged the anchor with our blend along a single line, and the model said that line
tops out at about 0.94658. Version 4 widens the search. Many engines in the garage are *exactly* the files their
authors submitted, so their public AUCs are known. For the public objective these are not estimates: a file's public
score **is** its AUC on the public rows, up to rounding. The Fisher model generalizes to any number of files:

$$\text{AUC}(v) = \Phi\!\left(\frac{v^\top\delta}{\sqrt{2\,v^\top\Sigma_w v}}\right),\qquad \Sigma_w = \Sigma - \pi(1-\pi)\,\delta\delta^\top,$$

where every pairwise correlation in $\Sigma$ is the top-half correlation mapped through the calibrated $k$ (Lap 10).
We maximize over weights $v \ge 0$ that sum to 1, with three guard rails: our OOF-validated blend keeps **at least
10%**, no other single file gets more than 30%, and each of the two jazivxt anchor files gets at most 90%.

**New in v6:** jazivxt's own notebook scored **0.94658** with a blend of the anchor and his single LightGBM (version 33).
The latest version of that notebook ships the same bytes, so we pin the file by its hash: if the author ever changes
it, the file drops out instead of entering with a score that belongs to other bytes.

**Input hygiene.** Before a notebook output enters the tower, its bytes are hashed. One of the new candidates, a
"teacher–student" notebook scored at 0.94656, turns out to be byte-for-byte identical to megayak's band-swapped file,
which was built by *reading* the public split. Its public score is real, but it says nothing about the private 80%,
so it is dropped.

🎓 **Pit-stop lesson: hash your inputs.** Two notebooks with different titles, authors and stories can ship the
same file. Titles describe intentions; hashes describe bytes.

🎓 **Pit-stop lesson: a public score is exact for the public objective and noisy for the private one.** Look at the
residuals of the OOF → public line below: BlamerX sits far above it. Some of that is a genuinely better view of the
data, and some is luck on 57k public rows. An empirical-Bayes split of the residual spread puts roughly two thirds on
"real". That is why our OOF-validated blend keeps a floor in the mix, and why the second output file uses labels only.

```python
anchor_path = find("submission_latest_best.csv", "zoom-zoom")
anchor = pd.read_csv(anchor_path).set_index(ID)[TARGET].reindex(test_ids).to_numpy(np.float64)
assert np.isfinite(anchor).all()
print(f"anchor sha256 {sha12(anchor_path)}… (7d46cc5b5dc1… is submission 56565120, public 0.94657)")
if not sha12(anchor_path).startswith("7d46cc5b5dc1"):
    print("WARNING: the anchor file changed since this notebook was measured; its public score may differ.")

hill_test = P_TEST @ W_FULL if HILL_ACCEPTED else P_TEST[:, ref].astype(np.float64)
own_test = (0.75 * zscore(hill_test) + 0.25 * zscore(stack_full.decision_function(P_TEST))
            if TEAM_ACCEPTED else hill_test)
g_anchor = probit(anchor).astype(np.float64)
g_own = probit(own_test).astype(np.float64)

lineage = np.array(LINEAGE)
TRANSFER = float(np.clip(np.polyfit(lineage[:, 0], lineage[:, 1], 1)[0], 0.0, 1.0))


def own_public(oof_auc, transfer=TRANSFER):
    # OOF → public transfer: v1's measured public score plus the paid part of the OOF gain since v1
    return OWN_V1_PUBLIC + transfer * (oof_auc - OWN_V1_OOF)


own_public_est = own_public(OWN_OOF)
print(f"OOF → public slope along the lineage: {TRANSFER:.2f} · our blend's public estimate {own_public_est:.6f}\n")

# rebuild the three measured challengers (v3's blend is this run's blend, which uses the same engines)
PAST = [("v1 challenger", {"fx_fusion": .375, "my_ensemble": .25, "nj_blend01": .125, "my_G_realmlp_3seed": .125,
                           "fx_C1_chinzo_lgbm": .125}, 0.5454829, 0.946449, 0.94655),
        ("v2 challenger", {"fx_fusion": .25, "my_ensemble": .125, "nj_blend01": .125, "my_G_realmlp_3seed": .125,
                           "fx_C1_chinzo_lgbm": .125, "bx_medvax_blamerx": .125, "lr_v19": .125}, 0.85, 0.946467, 0.94657),
        ("v3 challenger", None, 0.79, OWN_OOF, 0.94657)]
print(f"{'file':14s} {'w':>5s} {'top-half ρ':>11s} {'effective ρ':>12s} {'predicted':>10s} {'measured':>9s}")
for label, wts, w, own_oof_then, measured in PAST:
    g_then = g_own if wts is None else probit(
        sum(v * P_TEST[:, NAMES.index(k)].astype(np.float64) for k, v in wts.items())).astype(np.float64)
    r_eff = 1.0 - K_TOP * (1.0 - top_half_corr(g_anchor, g_then))
    print(f"{label:14s} {w:5.2f} {top_half_corr(g_anchor, g_then):11.6f} {r_eff:12.6f}"
          f" {blend_auc(w, r_eff, ANCHOR_PUBLIC, own_public(own_oof_then)):10.6f} {measured:9.5f}")


def engine(name):
    return P_TEST[:, NAMES.index(name)].astype(np.float64) if name in NAMES else None


def scored_file(hint):
    frame = pd.read_csv(find("submission.csv", hint))
    return probit(aligned(frame, test_ids, TARGET)).astype(np.float64)


# label, probit test vector of the exact scored file, its public AUC, its OOF AUC (None if not held)
TOWER = [("jazivxt anchor", g_anchor, ANCHOR_PUBLIC, None),
         ("our blend (estimate)", g_own, own_public_est, OWN_OOF),
         ("Fusion IV (mine)", engine("fx_fusion"), 0.94647, "fx_fusion"),
         ("BlamerX XGB", engine("bl_blamerx"), 0.94646, "bl_blamerx"),
         ("heuljax XGB", engine("hj_xgb"), 0.94638, "hj_xgb"),
         ("mizushima LGBM", engine("mz_lgb_comp"), 0.94636, "mz_lgb_comp"),
         ("aryankaisth HistGBM", engine("ak_histgbm"), 0.94636, "ak_histgbm"),
         ("yekenot RealMLP", engine("yk_realmlp_ptk"), 0.94621, "yk_realmlp_ptk"),
         ("evgendvorkin XGB", scored_file("single-xgb-cv-0-94607"), 0.94620, "ev_xgb"),
         ("tamerlan stack", scored_file("946072"), 0.94612, None),
         ("kirill0212 LGBM", engine("kr_lgb"), 0.94607, "kr_lgb"),
         ("jazivxt R model", "zoom-zoom-r-edition", 0.94621, None),
         ("mikhailnaumov XGB", "purchases-single-xgb", 0.94640, None),
         ("najiama LGBM v6", ("submission_LIGHTGBM.csv", "pure-lgbm"), 0.94638, None),
         ("tamerlan HiRGE blend", "custom-hirge-net", 0.94649, None),
         ("denpugovkin 15 teachers", "fifteen-teachers", 0.94644, None),
         ("crystalbaby teacher-student", "pseudo-labels", 0.94656, None),
         ("jazivxt V33 blend (anchor+LGBM)", ("blend.csv", "single-model-zoom-zoom"), 0.94658, None)]
# a notebook's score belongs to one version: pin the bytes of files whose score comes from an older version
PINNED = {"jazivxt V33 blend (anchor+LGBM)": "ecbcc81d4824"}

# input hygiene: hash every scored file that comes from a notebook output, and drop known public-split probes
PROBED = {"23052891f418": "megayak's band-swapped 0.94656 file (built by reading the public split)"}
checked, hygiene = [], []
for lab, vec, pub, o in TOWER:
    if isinstance(vec, (str, tuple)):
        pattern, hint = vec if isinstance(vec, tuple) else ("submission.csv", vec)
        path = find(pattern, hint)
        digest = sha12(path)
        if lab in PINNED and digest != PINNED[lab]:
            hygiene.append((lab, digest, f"dropped: not the scored file (expected {PINNED[lab]}…)"))
            continue
        if digest in PROBED:
            hygiene.append((lab, digest, "dropped: identical to " + PROBED[digest]))
            continue
        hygiene.append((lab, digest, "kept"))
        vec = probit(aligned(pd.read_csv(path), test_ids, TARGET)).astype(np.float64)
    if vec is not None:
        checked.append((lab, vec, pub, (single_auc[o] if isinstance(o, str) else o)))
TOWER = checked
print("input hygiene (sha256 of each notebook output):")
for lab, digest, verdict in hygiene:
    print(f"  {lab:31s} {digest}…  {verdict}")

# OOF -> public line across the single engines, and who beats it
pairs = np.array([(o, p) for lab, vec, p, o in TOWER[2:] if o is not None])
slope, icpt = np.polyfit(pairs[:, 0], pairs[:, 1], 1)
print(f"\nOOF → public across {len(pairs)} scored engines: slope {slope:.2f} (the lineage said {TRANSFER:.2f})")
scale = max(abs(p - (icpt + slope * o)) for lab, vec, p, o in TOWER[2:] if o is not None)
print("\n  residual from the OOF → public line (each # is one twelfth of the largest):")
for lab, vec, p, o in TOWER[2:]:
    if o is not None:
        r = p - (icpt + slope * o)
        n_hash = int(round(abs(r) / scale * 12))
        bar = (" " * (12 - n_hash) + "#" * n_hash + "|" + " " * 12) if r < 0 else (" " * 12 + "|" + "#" * n_hash + " " * (12 - n_hash))
        print(f"  {lab:20s} {bar}  {r:+.6f}   OOF {o:.6f} → public {p:.5f}")

m = len(TOWER)
R = np.eye(m)
for i in range(m):
    for j in range(i + 1, m):
        R[i, j] = R[j, i] = 1.0 - K_TOP * (1.0 - top_half_corr(TOWER[i][1], TOWER[j][1]))
d_sep = np.array([separation(t[2]) for t in TOWER])
S_w = R - PRIOR_POS_RATE * (1 - PRIOR_POS_RATE) * np.outer(d_sep, d_sep)


def tower_auc(v):
    v = np.asarray(v, dtype=np.float64)
    return norm.cdf(v @ d_sep / np.sqrt(2.0 * v @ S_w @ v))


ANCHORS = ("jazivxt anchor", "jazivxt V33 blend (anchor+LGBM)")
bounds = [(0.0, 0.90) if t[0] in ANCHORS else ((0.10, 1.0) if t[0].startswith("our blend") else (0.0, 0.30))
          for t in TOWER]
rng = np.random.default_rng(SEED)
best = None
for _ in range(30):
    x0 = rng.dirichlet(np.ones(m))
    res = minimize(lambda v: -1e3 * (v @ d_sep) / np.sqrt(2.0 * v @ S_w @ v), x0, method="SLSQP", bounds=bounds,
                   constraints=[{"type": "eq", "fun": lambda v: v.sum() - 1.0}], options={"ftol": 1e-12, "maxiter": 500})
    if res.success and (best is None or res.fun < best.fun):
        best = res
W_TOWER = np.clip(best.x, 0.0, None)
W_TOWER[W_TOWER < 0.005] = 0.0
W_TOWER /= W_TOWER.sum()
line_best = max(blend_auc(w, 1.0 - K_TOP * (1.0 - top_half_corr(g_anchor, g_own)), ANCHOR_PUBLIC, own_public_est)
                for w in np.arange(0.5, 0.901, 0.01))
PRED_PUBLIC = float(tower_auc(W_TOWER))

print("\n  P  file                             public   eff. ρ vs anchor  weight")
order = np.argsort([-t[2] for t in TOWER])
for pos, i in enumerate(order, 1):
    lab, vec, pub, o = TOWER[i]
    bar = "o=o>" if W_TOWER[i] > 0 else ""
    print(f" {pos:2d}  {lab:31s} {pub:.5f}{'*' if i == 1 else ' '}  {R[0, i]:.6f}        {W_TOWER[i]:5.1%} "
          f"{'=' * int(round(W_TOWER[i] * 40))}{bar}")
print("  * estimated")
print(f"\npredicted public AUC · anchor/own line (v1–v3 method): {line_best:.6f} · timing-tower mix: {PRED_PUBLIC:.6f}")

# predicted vs measured, every challenger this notebook has submitted
lab_ix = {t[0]: i for i, t in enumerate(TOWER)}
v4_mix = np.zeros(len(TOWER))
v4_mix[[lab_ix["jazivxt anchor"], lab_ix["our blend (estimate)"], lab_ix["BlamerX XGB"]]] = [0.6228, 0.15, 0.2272]
history = []
for label, wts, w, own_oof_then, measured in PAST:
    g_then = g_own if wts is None else probit(
        sum(v * P_TEST[:, NAMES.index(k)].astype(np.float64) for k, v in wts.items())).astype(np.float64)
    r_eff = 1.0 - K_TOP * (1.0 - top_half_corr(g_anchor, g_then))
    history.append((label.split()[0], blend_auc(w, r_eff, ANCHOR_PUBLIC, own_public(own_oof_then)), measured))
history += [("v4", float(tower_auc(v4_mix)), 0.94659), ("v5", float(tower_auc(v4_mix)), 0.94659),
            ("v6", PRED_PUBLIC, None)]
lo_x = min(min(h[1] for h in history), min(h[2] for h in history if h[2])) - 1e-5
hi_x = max(h[1] for h in history) + 1e-5
width = 60
pos = lambda v: int(round((v - lo_x) / (hi_x - lo_x) * (width - 1)))
print("\n  the pit-stop model versus the leaderboard   [===] measured score ± rounding   o model prediction")
for label, pred, meas in history:
    line = [" "] * width
    if meas is not None:
        for c in range(pos(meas - 5e-6), pos(meas + 5e-6) + 1):
            line[min(max(c, 0), width - 1)] = "="
        line[min(max(pos(meas - 5e-6), 0), width - 1)] = "["
        line[min(max(pos(meas + 5e-6), 0), width - 1)] = "]"
    line[min(max(pos(pred), 0), width - 1)] = "o"
    tail = f"pred {pred:.6f}" + (f" · measured {meas:.5f}" if meas else " · submitted by this run")
    print(f"  {label:3s} |{''.join(line)}| {tail}")
print(f"      {lo_x:.5f}{' ' * (width - 12)}{hi_x:.5f}")
pit(11, "timing tower solved")
```

## Lap 12 · Crossing the line

The final file is the timing-tower mix of the probit-scaled rankings, with the rules of the road applied once more
(the anchor and the single engines leak rows out of the hard blocks too). The OOF-only file gets the same rules.

```python
merged = sum(w * t[1] for w, t in zip(W_TOWER, TOWER) if w > 0)
final = apply_rules(merged, test) if RULES_ACCEPTED else rankdata(merged) / len(merged)
own_final = apply_rules(own_test, test) if RULES_ACCEPTED else rankdata(own_test) / len(own_test)
print(f"final vs anchor Spearman {np.corrcoef(rankdata(final), rankdata(anchor))[0, 1]:.6f}")
for label, mask, sign in rule_masks(test):
    print(f"  {label:40s} test rows {mask.sum():6,d} · anchor rank range {rankdata(anchor)[mask].min() / len(anchor):.4f}"
          f"–{rankdata(anchor)[mask].max() / len(anchor):.4f} → final {final[mask].min():.4f}–{final[mask].max():.4f}")


def save(values, name):
    frame = pd.DataFrame({ID: test_ids, TARGET: np.asarray(values, dtype=np.float64)})
    assert frame[ID].is_unique and len(frame) == len(sample) and np.isfinite(frame[TARGET]).all()
    assert frame[TARGET].between(0, 1).all()
    frame.to_csv(OUT / name, index=False)
    return sha12(OUT / name)


files = {
    "submission.csv": save(final, "submission.csv"),
    "submission_oof_only_hill.csv": save(own_final, "submission_oof_only_hill.csv"),
}
report = {
    "version": 6, "engines": NAMES, "fingerprint_dropped": bad, "medoids": medoids, "best_single": BEST_SINGLE,
    "best_single_oof_auc": float(single_auc.iloc[0]),
    "nested_hill_oof_auc": float(fast_auc(yb, hill_oof)), "hill_fold_wins": wins, "hill_accepted": bool(HILL_ACCEPTED),
    "team_car": {"accepted": bool(TEAM_ACCEPTED), "fold_gain_vs_hill": [float(g) for g in team_gain],
                 "oof_auc": float(fast_auc(yb, own_oof))},
    "hill_weights": weights.round(4).to_dict(), "folds": fold_table.to_dict("records"),
    "controls": race.to_dict("records"), "delong_delta": float(delta), "delong_ci95": [float(c) for c in ci],
    "rules": {"accepted": RULES_ACCEPTED, "folds": rule_table.to_dict("records"), "oof_auc_with_rules": OWN_OOF},
    "pit_stop_model": {"k_top50": K_TOP, "pairs": len(cal), "transfer_slope": TRANSFER,
                       "engine_slope": float(slope), "own_public_est": own_public_est},
    "timing_tower": {t[0]: {"public": t[2], "weight": float(w)} for t, w in zip(TOWER, W_TOWER)},
    "predicted_public_auc": PRED_PUBLIC, "predicted_public_line_only": float(line_best),
    "files": files, "runtime_seconds": round(time.time() - T0, 1),
    "which_file": {"submission.csv": "timing-tower mix (anchor + scored engines + our blend) + generator rules",
                   "submission_oof_only_hill.csv": "pure OOF-validated blend + generator rules (private-first)"},
}
(OUT / "grand_prix_report.json").write_text(json.dumps(report, indent=2, default=float))
pit(12, "checkered flag")

box("results · version 6", [
    f"predicted public AUC ........ {PRED_PUBLIC:.6f}",
    f"our blend OOF + rules ....... {OWN_OOF:.6f}",
    f"engines in the garage ....... {len(NAMES)}",
    f"runtime on CPU .............. {report['runtime_seconds']:.0f}s",
    "",
    "final mix                                   weight    public AUC",
] + [f"  {t[0]:40s} {w:6.1%}    {t[2]:.5f}{'*' if t[0].startswith('our') else ''}"
     for t, w in sorted(zip(TOWER, W_TOWER), key=lambda z: -z[1]) if w > 0] + [
    "",
    "* estimated from OOF · rules applied after the mix",
    "submission.csv = challenger · submission_oof_only_hill.csv = labels only",
])

top3 = [i for i in np.argsort(-W_TOWER) if W_TOWER[i] > 0][:3]
while len(top3) < 3:
    top3.append(None)
podium(*[(TOWER[i][0], f"{W_TOWER[i]:5.1%} of the mix · public {TOWER[i][2]:.5f}{'*' if i == 1 else ''}")
         if i is not None else ("-", "") for i in top3])
print(TROPHY)
print(f"nested hill OOF {report['nested_hill_oof_auc']:.6f} · team car {report['team_car']['oof_auc']:.6f} "
      f"(pole {report['best_single_oof_auc']:.6f}) · with rules {OWN_OOF:.6f} · predicted public {PRED_PUBLIC:.6f}"
      f" · {report['runtime_seconds']:.0f}s on CPU")
print({k: v for k, v in files.items()})
```

## 🏆 Podium and takeaways

1. **Probit ranks** put 83 differently-scaled engines on one track without changing any AUC.
2. **Correlation fingerprints** catch misaligned test files before they cost you a submission.
3. **More sources, more diversity.** heuljax's XGBoost and BlamerX's window encodings lifted the nested hill climb
   from 0.946461 (v2) to 0.946477. Both are new *feature views*, and on the public board such engines pay more per OOF
   point than blends do.
4. A **nested, gated hill climb** beats every stand-alone combiner: equal means, medoids, logistic and LightGBM
   stackers, label-based Fisher/LDA. A quarter of a logistic stack on top of it still adds a little on every fold.
5. **Rules of the road**: four huge, pure generator cells, validated fold by fold, are free AUC. Mined small cells are noise.
6. **The pit-stop model.** The binormal Fisher model overestimates blending gains when you feed it the plain
   correlation, even on labeled data. The correlation inside the top half of the ranking, scaled by one constant
   learned on 66 labeled pairs, predicts all three measured challengers of this notebook within 0.00001.
7. **The timing tower.** The same calibrated model, run over every file whose exact public score is known, finds a
   better mix than the anchor/own line of versions 1–3. It puts real weight on BlamerX, the engine that sees the data
   most differently from the anchor.
8. **Hash your inputs.** A new 0.94656 "teacher–student" file was byte-identical to a public-split probe. Titles
   describe intentions; hashes describe bytes.
9. **Pin what you did not make.** A notebook's score belongs to one version. Pin the bytes by hash, or the score
   you read can silently start describing a different file.
10. **Know which leaderboard you are optimizing.** Public scores are exact for the public 20% and noisy for the other
   80%. The challenger uses them. The OOF-only file doesn't. Keep one of each as your two final picks.

**Two files, two purposes.** `submission.csv` is the challenger (submitted from this notebook, so its score is
shown on this page). `submission_oof_only_hill.csv` uses labels only.
